```
import pandas as pd
import numpy as np
import matplotlib.pyplot as pt
```

```
score = [45,55,87,55,44,58,98,54,78,53,68,96,
        54,78,98,65,33,87,54,87,55,85,58,54,57,
        40,65,52,44,66,40,98,54,57,41,25,68,58]

df = pd.DataFrame({'Score' : score})
df.head()
```

```
df.describe()
```

```
mean_val = df['Score'].mean()
std_val = df['Score'].std()
count_val = df['Score'].count()
min_val = df['Score'].min()
max_val = df['Score'].max()
mode_val = df['Score'].mode()
median_val = df['Score'].median()

print('mode :' ,mode_val)
print('median :' ,median_val)
print('count :', count_val)
print('mn :' ,min_val)
print('mx :',max_val)
print( 'std :' ,std_val)
print(  'mean :' ,mean_val)
```

```
bins = [39,49,59,69,79,89,100]
labels = ['40-49','50-59','60-69','70-79','80-89','90-100']
df['Grade'] = pd.cut(df['Score'],bins=bins,labels=labels)

fre_table = df["Grade"].value_counts().sort_index()

print(fre_table)
```
